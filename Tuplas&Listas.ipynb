**Aula de Listas & Tuplas**

**Tuplas** são estruturas que armazenam dados que podem ser heterogêneas mas que quando criadas não podem ter seu conteúdo alterado, ou seja, propriedade de IMUTABILIDADE

**Listas**

são estruturas que armazenam dados que podem ser heterogêneas mas que tem seus dados e dimensão alterados conforme as necessidades, ou seja. **MUTÁVEL**

```python
'''
Programa para demonstrar a utilização de Tuplas
Autor: Jonatas.L
Data: 11 de Março de 2026
'''
dias = ('Seg', 'Ter', 'Qua', 'Quin', 'Sex', 'Sab', 'Dom')
print("Conteúdo da Tupla: ")
#type(dias) - exibir o tipo da váriavel da estrutura
tamanho = len(dias)
for i in dias :
  print(i)
```

```python
'''
Programa para demonstrar a utilização de Listas
Autor: Jonatas.L
Data: 11 de Março de 2026
'''

nomes = ['José','Pedro','Carla','Betania','Mauro']
print("Conteúdo da Lista: ")
print(nomes)
# Adicionar um elemento na estrutura Lista
nomes.append('Ezequiel')
# Com o nome Ezequiel adicionado:
print(nomes)
print(f'Tamanho da Lista: {len(nomes)} elementos')
# Excluindo Pessoa da Lista:
nomes.remove('Carla')
print(nomes)
print(f'Tamanho da Lista: {len(nomes)} elementos')
print(nomes[2])
nomes[2] = 'Ptolomeu'
print(nomes)
tamanho = len(nomes)
for i in range(1, tamanho, 2): #Start, stop-1 ,step
  print(nomes[i])
```
